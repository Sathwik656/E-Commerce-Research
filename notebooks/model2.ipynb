# Task 5 - Construct Spending Features + Target

```python
max_olist_date = olist_clean["order_purchase_timestamp"].max()

spending_observation_end = (
    max_olist_date - pd.Timedelta(days=90)
)

print("Maximum purchase date:", max_olist_date)
print(
    "Spending observation period ends:",
    spending_observation_end
)
```

```python
olist_model = olist_clean[
    olist_clean["order_status"] == "delivered"
].copy()

print("Delivered orders:", len(olist_model))
print(
    "Unique customers:",
    olist_model["customer_unique_id"].nunique()
)
```

```python
print(
    olist_model.isnull().sum()
)
```

```python
required_columns = [
    "customer_unique_id",
    "order_purchase_timestamp",
    "payment_value",
    "number_of_items",
    "number_of_products",
    "number_of_sellers",
    "total_item_price",
    "total_freight_value",
    "payment_installments"
]

olist_model = olist_model.dropna(
    subset=required_columns
).copy()
```

```python
print(
    "Rows after removing incomplete modelling records:",
    len(olist_model)
)
```

```python
historical_olist = olist_model[
    olist_model["order_purchase_timestamp"]
    <= spending_observation_end
].copy()
```

```python
future_olist = olist_model[
    olist_model["order_purchase_timestamp"]
    > spending_observation_end
].copy()
```

```python
print(
    "Historical orders:",
    len(historical_olist)
)

print(
    "Future orders:",
    len(future_olist)
)
```

```python
print(
    "Historical period:",
    historical_olist["order_purchase_timestamp"].min(),
    "to",
    historical_olist["order_purchase_timestamp"].max()
)

print(
    "Future period:",
    future_olist["order_purchase_timestamp"].min(),
    "to",
    future_olist["order_purchase_timestamp"].max()
)
```

```python
historical_olist["OrderValue"] = (
    historical_olist["payment_value"]
)
future_olist["FutureOrderValue"] = (
    future_olist["payment_value"]
)
```

```python
customer_last_purchase = (
    historical_olist
    .groupby("customer_unique_id")["order_purchase_timestamp"]
    .max()
)
recency = (
    spending_observation_end
    - customer_last_purchase
).dt.days
```

```python
recency = (
    recency
    .rename("Recency")
    .reset_index()
)
```

```python
frequency = (
    historical_olist
    .groupby("customer_unique_id")["order_id"]
    .nunique()
    .rename("Frequency")
    .reset_index()
)
```

```python
monetary = (
    historical_olist
    .groupby("customer_unique_id")["OrderValue"]
    .sum()
    .rename("HistoricalSpending")
    .reset_index()
)
```

```python
total_items = (
    historical_olist
    .groupby("customer_unique_id")["number_of_items"]
    .sum()
    .rename("TotalItems")
    .reset_index()
)
```

```python
total_products = (
    historical_olist
    .groupby("customer_unique_id")["number_of_products"]
    .sum()
    .rename("TotalProducts")
    .reset_index()
)
```

```python
total_sellers = (
    historical_olist
    .groupby("customer_unique_id")["number_of_sellers"]
    .sum()
    .rename("TotalSellers")
    .reset_index()
)
```

```python
average_order_value = monetary.copy()

average_order_value["AverageOrderValue"] = (
    average_order_value["HistoricalSpending"]
    / frequency["Frequency"]
)

average_order_value = average_order_value[
    [
        "customer_unique_id",
        "AverageOrderValue"
    ]
]
```

```python
average_freight = (
    historical_olist
    .groupby("customer_unique_id")["total_freight_value"]
    .mean()
    .rename("AverageFreightValue")
    .reset_index()
)
```

```python
average_installments = (
    historical_olist
    .groupby("customer_unique_id")["payment_installments"]
    .mean()
    .rename("AverageInstallments")
    .reset_index()
)
```

```python
spending_features = recency.copy()
spending_features = spending_features.merge(
    frequency,
    on="customer_unique_id",
    how="left"
)

spending_features = spending_features.merge(
    monetary,
    on="customer_unique_id",
    how="left"
)

spending_features = spending_features.merge(
    total_items,
    on="customer_unique_id",
    how="left"
)

spending_features = spending_features.merge(
    total_products,
    on="customer_unique_id",
    how="left"
)

spending_features = spending_features.merge(
    total_sellers,
    on="customer_unique_id",
    how="left"
)

spending_features = spending_features.merge(
    average_order_value,
    on="customer_unique_id",
    how="left"
)

spending_features = spending_features.merge(
    average_freight,
    on="customer_unique_id",
    how="left"
)

spending_features = spending_features.merge(
    average_installments,
    on="customer_unique_id",
    how="left"
)
```

```python
future_spending = (
    future_olist
    .groupby("customer_unique_id")["FutureOrderValue"]
    .sum()
    .rename("FutureSpending")
    .reset_index()
)
```

```python
spending_features = spending_features.merge(
    future_spending,
    on="customer_unique_id",
    how="left"
)
```

# Task 6.0 Fixing errors

```python
max_olist_date = olist_model["order_purchase_timestamp"].max()

print("Maximum Olist purchase date:", max_olist_date)
```

```python
def evaluate_future_window(days):
    cutoff = max_olist_date - pd.Timedelta(days=days)

    historical = olist_model[
        olist_model["order_purchase_timestamp"] <= cutoff
    ]

    future = olist_model[
        olist_model["order_purchase_timestamp"] > cutoff
    ]

    historical_customers = historical["customer_unique_id"].unique()

    future_spending = (
        future
        .groupby("customer_unique_id")["payment_value"]
        .sum()
    )

    future_spending = future_spending.reindex(
        historical_customers,
        fill_value=0
    )

    positive_customers = (future_spending > 0).sum()
    total_customers = len(future_spending)
    zero_customers = (future_spending == 0).sum()

    return {
        "Future Window (Days)": days,
        "Cutoff Date": cutoff,
        "Historical Customers": total_customers,
        "Positive Future Spending": positive_customers,
        "Zero Future Spending": zero_customers,
        "Positive %": round(
            positive_customers / total_customers * 100,
            2
        ),
        "Zero %": round(
            zero_customers / total_customers * 100,
            2
        )
    }    
```

```python
window_results = pd.DataFrame([
    evaluate_future_window(90),
    evaluate_future_window(120),
    evaluate_future_window(180),
    evaluate_future_window(270),
    evaluate_future_window(365)
])

window_results
```

```python
print(
    window_results[
        [
            "Future Window (Days)",
            "Historical Customers",
            "Positive Future Spending",
            "Positive %",
            "Zero %"
        ]
    ]
)
```

```python
def future_spending_distribution(days):
    cutoff = max_olist_date - pd.Timedelta(days=days)

    historical = olist_model[
        olist_model["order_purchase_timestamp"] <= cutoff
    ]

    future = olist_model[
        olist_model["order_purchase_timestamp"] > cutoff
    ]

    historical_customers = historical["customer_unique_id"].unique()

    future_spending = (
        future
        .groupby("customer_unique_id")["payment_value"]
        .sum()
        .reindex(historical_customers, fill_value=0)
    )

    return future_spending
```

```python
print(
    "Total customers:",
    customer_order_counts.shape[0]
)

print(
    "Customers with 2+ orders:",
    (customer_order_counts >= 2).sum()
)

print(
    "Customers with only 1 order:",
    (customer_order_counts == 1).sum()
)
```

```python
next_order_data["PreviousOrderCount"] = (
    next_order_data
    .groupby("customer_unique_id")
    .cumcount()
)
```

```python
next_order_data["NextOrderSpending"] = (
    next_order_data["payment_value"]
)
```

```python
next_order_data["CumulativeSpending"] = (
    next_order_data
    .groupby("customer_unique_id")["payment_value"]
    .cumsum()
)
```

```python
next_order_data["HistoricalSpending"] = (
    next_order_data["CumulativeSpending"]
    - next_order_data["payment_value"]
)
```

```python
next_order_data["CumulativeItems"] = (
    next_order_data
    .groupby("customer_unique_id")["number_of_items"]
    .cumsum()
)

next_order_data["TotalItems"] = (
    next_order_data["CumulativeItems"]
    - next_order_data["number_of_items"]
)
```

```python
next_order_data["CumulativeProducts"] = (
    next_order_data
    .groupby("customer_unique_id")["number_of_products"]
    .cumsum()
)

next_order_data["TotalProducts"] = (
    next_order_data["CumulativeProducts"]
    - next_order_data["number_of_products"]
)
```

```python
next_order_data["CumulativeSellers"] = (
    next_order_data
    .groupby("customer_unique_id")["number_of_sellers"]
    .cumsum()
)

next_order_data["TotalSellers"] = (
    next_order_data["CumulativeSellers"]
    - next_order_data["number_of_sellers"]
)
```

```python
next_order_data["AverageOrderValue"] = (
    next_order_data["HistoricalSpending"]
    / next_order_data["PreviousOrderCount"]
)
```

```python
next_order_data["CumulativeFreight"] = (
    next_order_data
    .groupby("customer_unique_id")["total_freight_value"]
    .cumsum()
)

next_order_data["HistoricalFreight"] = (
    next_order_data["CumulativeFreight"]
    - next_order_data["total_freight_value"]
)
```

```python
next_order_data["AverageFreightValue"] = (
    next_order_data["HistoricalFreight"]
    / next_order_data["PreviousOrderCount"]
)
```

```python
next_order_data["CumulativeInstallments"] = (
    next_order_data
    .groupby("customer_unique_id")["payment_installments"]
    .cumsum()
)

next_order_data["HistoricalInstallments"] = (
    next_order_data["CumulativeInstallments"]
    - next_order_data["payment_installments"]
)
```

```python
next_order_data["AverageInstallments"] = (
    next_order_data["HistoricalInstallments"]
    / next_order_data["PreviousOrderCount"]
)
```

```python
next_order_data["PreviousPurchaseDate"] = (
    next_order_data
    .groupby("customer_unique_id")[
        "order_purchase_timestamp"
    ]
    .shift(1)
)
```

```python
next_order_data["Recency"] = (
    next_order_data["order_purchase_timestamp"]
    - next_order_data["PreviousPurchaseDate"]
).dt.total_seconds() / (24 * 60 * 60)
```
